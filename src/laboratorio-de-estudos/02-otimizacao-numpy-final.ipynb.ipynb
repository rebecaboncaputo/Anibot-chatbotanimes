# Etapa 2: usando NumPy para trabalhar com vetores

### Uso do numpy

Na etapa anterior, todos os calculos foram feitos manualmente, ex:
- abs(vetor_a[0] - vetor_b[0])
- abs(vetor_a[1] - vetor_b[1])
- abs(vetor_a[2] - vetor_b[2])

Isso funcionou porque nossos vetores tinham apenas três números. Mas um embedding real pode ter centenas de dimensões. Escrever uma expressão para cada posição seria impraticável.
O NumPy é uma biblioteca Python especializada em:
trabalhar com grandes conjuntos de números;
realizar operações matemáticas sobre vetores e matrizes;
executar cálculos de forma mais curta e eficiente;
fornecer funções prontas para produto escalar, norma e outras operações.

Nessa etapa vou estar refazendo os códigos para adicionar o numpy e explicando as função numpy usadas. A lógica dos códigos sao as mesmas usadas no primeiro notebook.

### Primeiro código: Transformando lista em arrays

```python
import numpy as np 

pedido = np.array([2,4,6])
anime_a = np.array([3,4,7])

print("Pedido: ", pedido)
print("Anime A: ", anime_a)
```

#### O que esse código faz?
Nesse código é possível entender o conceito básico de transfomação de vetores (embeddings) em numpy:
- np.array é uma função pronta importada do numpy, que tranforma uma lista em um array.

### Segundo código: Subtração de vetores sem repetir posições

```python
import numpy as np 

pedido = np.array([2,4,6])
anime_a = np.array([3,4,7])

print("Pedido: ", pedido)
print("Anime A: ", anime_a)

diferenca = np.sum(np.abs(pedido - anime_a))
print("Diferença absoluta:", diferenca)
```

#### O que esse código faz?

Nesse código podemos começar a observar como a biblioteca numpy trás mais praticidade o código. Por mais que não seja exclusivo do numpy, usando ele subtrair listas diretamente em vez de ter que escrever elemento por elemento.

- A variável "diferenca" representa a diferença entre as duas listas
- A função np.abs é uma função numpy que tranforma o valor, em um valor absoluto.
- A função np.sum é uma função numpy que soma valores.

### Quarto código: Produto escalar

```python
import numpy as np
 
vetor_a = np.array([2,4,6])
vetor_b = np.array([3,4,7])

produto_escalar = np.dot(vetor_a, vetor_b)

print("Produto escalar:", produto_escalar)
```

#### O que esse códiigo faz?

Para entender  similaridade do cosceno, é preciso entender em etapas sua fórmula, contendo nela o produto escalar
- A função np.dot calcula o vetor escalar entre dois vetores. Essa função faz automaticamente: multiplica os valores correspondentes; soma os resultados; devolve um único número.


### Quinto código: Comprimento matemático do vetor

```python
import numpy as np
 
vetor_a = np.array([2,4,6])
vetor_b = np.array([3,4,7])

Tamanho_vetor_a =  np.linalg.norm(vetor_a)
Tamanho_vetor_b =  np.linalg.norm(vetor_b)

print("Tamanho vetor A: ",Tamanho_vetor_a )
print("Tamanho vetor B: ",Tamanho_vetor_b )
```

#### O que esse códiigo faz?

Aqui segue a próxima etapa para o entendimento para a formula de similaridade de cosceno: Norma. Norma é o cálculo que realizamos para saber o tamanho de um vetor.
- np.linalg é um módulo do NumPy voltado para álgebra linear.
- A função np.linalg.norm(vetor_a) calcula o comprimento do vetor.


### Sexto código: Calculando a similaridade do cosseno

```python
import numpy as np

def similaridade_cosceno(vet1,vet2):
    produto_escalar = np.dot(vet1, vet2)
    Tamanho_vetor1 =  np.linalg.norm(vet1)
    Tamanho_vetor2 =  np.linalg.norm(vet2)
    
    return produto_escalar / (Tamanho_vetor1 * Tamanho_vetor2)
    
```

#### O que esse código faz?
Esse código é a fórmula da similaridade do cosceno, onde juntei todos os coneitos revisados até agora.

# Codigo final: Aplicação de emplendings 

```python
import numpy as np


def similaridade_cosseno(vetor_a, vetor_b):
    produto_escalar = np.dot(vetor_a, vetor_b)

    norma_a = np.linalg.norm(vetor_a)
    norma_b = np.linalg.norm(vetor_b)

    return produto_escalar / (norma_a * norma_b)


pedido = np.array([7, 2, 6])

anime_a = np.array([8, 2, 7])
anime_b = np.array([7, 3, 6])
anime_c = np.array([1, 9, 2])

similaridade_a = similaridade_cosseno(pedido, anime_a)
similaridade_b = similaridade_cosseno(pedido, anime_b)
similaridade_c = similaridade_cosseno(pedido, anime_c)

print("Similaridade com Anime A:", similaridade_a)
print("Similaridade com Anime B:", similaridade_b)
print("Similaridade com Anime C:", similaridade_c)
```

#### O que esse código faz?

Este código representa a fundação matemática de um sistema de recomendação baseado em embeddings e busca semântica, utilizando a biblioteca NumPy para calcular a similaridade do cosseno entre um vetor de pedido e diferentes vetores de animes. Por meio de vetores manuais e operações que determinam o produto escalar e as normas, o script traduz a geometria espacial em pontuações numéricas de proximidade, demonstrando de forma prática como a lógica vetorial avalia a compatibilidade de significado entre textos.
