# Etapa 1: vetores artificiais em Python puro


### Primeiro código: criando vetores


```python
pedido = [7,3,6]

anime_a = [8,2,7]
anime_b = [6,4,5]
anime_c = [3,9,1]

print("Pedido: ", pedido)
print("Anime A", anime_a)
print("Anime B", anime_b)
print("Anime C", anime_c)
```

#### O que esse código faz?
Nesse código é possível entender o conceito básico de vetores (embeddings):
- A lista pedido representa o vetor numérico (embedding) com as preferências de um usuário.
- As outras 3 listas (anime_a, anime_b, anime_c) representam os animes cadastrados no dataset.
- Na IA transformamos gostos, palavras e itens em listas de números para que o computador consiga calcular a afinidade entre eles.

### Segundo Código: Acessando posições do vetor


```python
pedido = [7,3,6]

anime_a = [8,2,7]
anime_b = [6,4,5]
anime_c = [3,9,1]

print("Pedido: ", pedido)
print("Anime A", anime_a)
print("Anime B", anime_b)
print("Anime C", anime_c)

print("Primeira caracteristica do pedido: ", pedido[0])
print("Segunda caracteristica do pedido: ", pedido[1])
print("Terceira caracteristica do pedido: ", pedido[2])
```

#### O que esse código faz?
Neste código, acessamos índices individuais das listas (pedido[0], pedido[1] e pedido[2]).

Na lógica de Inteligência Artificial, isso simboliza:
- Dimensões / Features (Características do Embedding): Cada posição do vetor não é apenas um número genérico, mas representa uma dimensão semântica mensurável. Por exemplo:
  - pedido[0] (índice 0): Intensidade de Ação (nota 7)
  - pedido[1] (índice 1): Intensidade de Comédia (nota 3)
  - pedido[2] (índice 2): Intensidade de Romance (nota 6)
- Extração de Atributos: Para comparar itens no espaço multidimensional, a IA precisa ler cada dimensão individualmente e ponderar o valor de cada atributo.

### Terceiro Código: A primeira comparação: distância por posição


```python
def distancia_simples(vetor_a, vetor_b):
    diferenca_1 = abs(vetor_a[0] - vetor_b[0])
    diferenca_2 = abs(vetor_a[1] - vetor_b[1])
    diferenca_3 = abs(vetor_a[2] - vetor_b[2])
    return diferenca_1 + diferenca_2 + diferenca_3

pedido = [7, 3, 6]

anime_a = [8, 2, 7]
anime_b = [6, 4, 5]
anime_c = [3, 9, 1]

distancia_a = distancia_simples(pedido, anime_a)
distancia_b = distancia_simples(pedido, anime_b)
distancia_c = distancia_simples(pedido, anime_c)

print("Distancia até o Anime A: ", distancia_a)
print("Distancia até o Anime B: ", distancia_b)
print("Distancia até o Anime C: ", distancia_c)
```

#### O que esse código faz?
Aqui criamos a nossa primeira função de similaridade vetorial (distancia_simples).

Na lógica de Inteligência Artificial, isso simboliza:
- Distância de Manhattan ($L_1$ Norm): Uma métrica matemática clássica utilizada por algoritmos de IA para medir quão semelhantes ou distintos dois itens são no espaço vetorial.
- O papel do abs() (Módulo / Valor Absoluto): A divergência entre preferências não pode ser negativa (não existe erro de -2). O abs() calcula a distância real entre os valores em cada dimensão:
  - Anime A: $|7 - 8| + |3 - 2| + |6 - 7| = 1 + 1 + 1 = 3$
  - Anime B: $|7 - 6| + |3 - 4| + |6 - 5| = 1 + 1 + 1 = 3$
  - Anime C: $|7 - 3| + |3 - 9| + |6 - 1| = 4 + 6 + 5 = 15$
- Princípio Fundamental da IA (Menor Distância = Maior Afinidade):
  - Anime A e Anime B (Distância 3): Têm altíssima proximidade semântica com o pedido do usuário.
  - Anime C (Distância 15): É despriorizado pela IA por divergir fortemente do perfil desejado.
